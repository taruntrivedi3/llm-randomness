# NumPy baseline

The histogram recipe used in the article, extracted from `nn_tutorial.ipynb`. Running this cell generates a new NumPy baseline, not the exact original article values. The four LLM notebooks analyze recorded replies.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
s = np.random.normal(loc=10, scale=2, size=2000)
with plt.style.context("ggplot"):
    plt.figure(figsize=(9, 5))
    plt.hist(s, bins=50, color="#0891b2", edgecolor="white", alpha=0.85)
    plt.xlabel("Value")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()
```
